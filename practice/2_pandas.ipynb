### Creating pandas data structures

pandas has two main data structures:

1.  [`Series`](https://pandas.pydata.org/docs/reference/api/pandas.Series.html):
    observations of a *single* variable.
2.  [`DataFrame`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html):
    container for *several* variables.

*Example: Create Series from 1-dimensional NumPy array*

```python
import numpy as np
import pandas as pd

# Create an array containing integers from 5 to 9
data = np.arange(5,10)

# Convert the array into a pandas series
s = pd.Series(data)

print(s)

# Results:
# Left side: index labels, automatically assigned from 0
# Right side: the actual values 
# dtype: int64 means that the values are stored as integers
```

*Example: Create DataFrame from NumPy array*

-   We can create 2-dimensional arrays from 1-dimensional ones with 
    [`reshape()`](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html)
- reshape(rows, columns) 

```python
# Create integers from 0 to 14, arranged into an array/matrix with 3 columns

    # Can alternatively use reshape(-1, number of columns)...
    # ...the "-1" let's NumPy calculate the required number of rows

data = np.arange(15).reshape(5, 3)
print(data)

# Choose column names
variable_names = ['A', 'B', 'C']

# Create and display a table

    # pd.DataFrame() creates a pandas table from data you provide 

df = pd.DataFrame(data, columns = variable_names)
print(df)
```

*Example: Create DataFrame with non-homogeneous data from dictionary*

```python
# Names: text values
names = ['Alice', 'Bob']

# Convert date strings into pandas datetime values
birthdates = pd.to_datetime(['1985-01-01', '1997-05-12'])

# Incomes: np.nan represents a missing value
incomes = np.array([600000, np.nan])

# Create a table using a dictionary

    # The dictionary connects each 'column name' to its data 
df = pd.DataFrame({'Name': names, 'Birthdate': birthdates, 'Incomes': incomes})
df
```

## Indexing

pandas supports two types of indexing:

1.  Indexing by position (same as Python containers and NumPy arrays)
2.  Indexing by label, i.e., by the values assigned to the row or column *index*.

pandas indexing is performed either by using brackets `[]`, or by using
[`.loc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html)
for label indexing, or
[`.iloc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html)
for positional indexing.

Indexing via `[]` can be somewhat confusing:

-   specifying `df['name']` returns the column `name` as a `Series` object.
-   specifying a range such as `df[5:10]` returns the *rows*
    associated with the *positions* 5,...,9.

    **Recommendation:** Don't use this, there are less confusing ways to select rows.

*Example: Selecting a single column*

-   Select column `'Name'` from Titanic data set

```python
df['Name']

# This returns a Series: one labelled column.
```

*Example: Selecting multiple columns*

-   Select columns `'Name'` and `'Sex'` from Titanic data set
-   Need to specify multiple columns as `list`

```python
df[['Name', 'Sex']]

# This returns a DataFrame: a table containing the selected columns 
# The double-brackets have two seperate jobs:
    # 1. ['Name', 'Sex] creates a list of column names
    # 2. df[...] uses that list to select columns
```

### Creating and manipulating indices


Three main methods to create/manipulate indices:

1.   Create a new `Series` or `DataFrame` object with a custom index
    using the `index` argument.
2.   [`set_index(keys=['column1', ...])`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.set_index.html)
    uses the values of `column1`
    and optionally additional columns as indices, discarding the current index.
3.   [`reset_index()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.reset_index.html) 
    resets the index to its default value, a sequence
    of increasing integers starting at 0.

#### Creating custom indices

*Example: Create `Series` with custom index*

```python
import pandas as pd

s = pd.Series([0, 1, 2], index = ['a', 'b', 'c'])
s

# The index row by default is 0, 1, 2, 3, ..., but we can choose other labels.
# index = ['a', 'b', 'c'] replaces digits with these letters. The first element is 'a' while its  position is 0.
```

#### Manipulating indices

-   Default: return a new object; leave the original unchanged.
-   Assign the result back, or use `inplace=True`.
-   With `inplace=True`, both methods return `None`.
-   `set_index()`: `DataFrame` only; `reset_index()`: also `Series`.

*Example: Set DataFrame index from column*

-   Use the
    [`set_index()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.set_index.html)
    method
-   Optionally specify `append=True` to add as *additional* index levels

```python
df = pd.DataFrame({'A': [10, 20, 30], 'B': ['a','b','c']})

df
```

```python
# We can set column B and its values as the row labels 
# Column B becomes the index column

df2 = df.set_index('B')

df2

# The original df is unchanged. We stored our new table with column B in the index column under df2.
```

```python
# Method 1

# By using "inplace = True", we change the existing table df.
# Now df will have column B in the index column.

# Run this cell after the first cell, and not the second, in order for it to work. 

df.set_index('B', inplace=True)

# Since inplace = True returns None, it sends no table back. In order to display the new index, you can display it with df. 

df

# Now df itself uses column B as its index:
```

```python
# Method 2
# Alternatively, to permanently change the index with a column, we can also do this:

df = df.set_index('B')

df
```

```python
# When using method 1, inplace = True, do NOT write this:

df = df.setindex('B', inplace = True)

# Python does two things:
# 1. Changes the existing table’s index.
# 2. Assigns the returned value — None — to df.
# Now df refers to None, rather than your table.

# Either use:
df.set_index('Name', inplace=True)  # Change df directly. DO NOT STORE IT IN df.
df = df.set_index('Name')           # Return the updated table and store it in df.
```

```python
# append=True keeps the current index and adds another index level

df = pd.DataFrame({'A': [10, 20, 30], 'B': ['a','b','c']})

df2 = df.set_index('B', append=True)
df2

# You now have two index levels: the original numbers and the letters from B. Together, they form a MultiIndex.
# Without append=True, the letters replace the numbered index (as we saw earlier).
```

append=True controls whether the old index is kept. 

inplace=True controls whether the original DataFrame is changed

*Example: Reset DataFrame index*

-   Use the
    [`reset_index()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.reset_index.html)
    method
-   Use `drop=True` to discard the old index; otherwise, it becomes a column.

```python
# To restore the index to default (remove column B and have indices instead)

df = pd.DataFrame({'A': [10, 20, 30], 'B': ['a','b','c']})

df.set_index('B', inplace=True)
```

```python
df.reset_index()

# This preserves the old index as a column.
```

```python
# To discard the old labels instead, do this:

df = df.reset_index(drop = True)

df

# This displays only A, discarding column B and its letters.
```

### Selecting elements

Recommended rules for indexing:

1.  Use `df['name']` to select *columns*.
2.  Use
    [`.loc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html)
    to select by label.
3.  Use
    [`.iloc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html)
    to select by position.

*Demo data set used for this section:*

```python
# Create demo data with 3 columns and 5 rows

# Column labels
columns = ['X', 'Y', 'Z']
# Row labels
rows = ['a', 'b', 'c', 'd', 'e']

values = np.arange(len(rows))

# Create data dictionary using a list comprehension
data = {col: [f'{col}{val}' for val in values] for col in columns}

# Create DataFrame from dictionary
df = pd.DataFrame(data, index=rows)

df
```

**Selection by label**

-   Use
    [`.loc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html)
    to select rows and/or columns *by label*
-   Can use *slicing* where last element is *included*

The general structure is: df.loc[row_label, column_label]

```python
# This selects row labelled 'b'

df.loc['b']
```

```python
# To select a single value, we can select a row and and column to locate it

df.loc['b', 'X']
```

```python
# For multiple rows and columns, use lists:

df.loc[['b','d'],['X','Z']]
```

```python
# Another method is to display multiple rows and columns is using the ":" to define a range
# 'a':'c' selects rows from a-c, and columns X to Y

df.loc['a':'c','X':'Y']
```

**Selection by position**

-   Use
    [`.iloc[]`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html)
    to select rows and/or columns *by integer position*
-   Reset the demo index to default integers before the exercise.

```python
# .iloc[] counts positions from zero, like NumPy

df.iloc[1,0] 

# This selects the second row (index 1), and the first column, giving X1.
```

```python
# This selects the first four rows and first two columns (index 0 and 1)

df.iloc[0:4, 0:2]
```

Unlike a .loc label slice, an .iloc slice excludes its endpoint.

| Code | Selects using | Endpoint in a slice |
|---|---|---|
| `df.loc['b':'d']` | Row labels | Includes `d` |
| `df.iloc[1:3]` | Row positions | Excludes position `3` |
