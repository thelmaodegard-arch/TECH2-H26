# NumPy practice

NumPy is a Python package for working with numerical data.
An array holds values that we can select and calculate with.

```python
import numpy as np

# Create a Python list, stored under the name "values"
values = [2, 3, -1, 7, 4]

# Create a NumPy array from that list, and store it under the name "numbers"
numbers = np.array(values)

print(values)

print(numbers)

# Our results show two lines:
# 1. the first is a line
# 2. the second is an array (the missing commas are how NumPy displays arrays)
```

Multiplying a list by 2 repeats the list twice. Multiplying a NumPy array by 2 doubles each value.

```python
# Now let's multiply the list and the array with 2

print(values * 2)
print(numbers * 2)

# Here the results show:
# List: the list repeats
# Array: a NumPy array multiplies each element 
```

Next, let’s select individual values. Both lists and arrays use positions called indices, starting at 0:

```python
# Numbers (array) = [2, 3, -1, 7, 4]

print(numbers[0]) #2
print(numbers[3]) #7
print(numbers[-1]) #4

# Alternative method to access indexes and values: 4
# 1. Use -1 to select the last element (counting from the end), likewise -2 will select 7
# 2. Use 4 to select the last element from start
```

Now we can connect indexing to Richard’s lecture example: finding where the largest value is.

```python
# np.argmax(numbers) returns the INDEX of the largest value, stored in "imax"
imax = np.argmax(numbers)

# numbers[imax] retrieves the VALUE at that index, stored in "vmax"
vmax = numbers[imax]

print(imax)
print(vmax)

# So index [3] has the largest value, and is stored in imax
# By assigning the index with numbers array, vmax will show the value for index [3]

# imax uses argmax to store the index of the largest value in numbers.
# numbers[imax] selects the value at that index in the numbers array.
# vmax stores that selected value.
```

argmax tells me where the largest value is. Indexing with that result tells me what the value is.
1. argmax returns the index of the largest value → 3.
2. numbers[imax] retrieves the value at that index → 7.

Let's try to find both the biggest and smallest value with a new array:

```python
# Try a new array

# What is the biggest value and its index? 

numbers = np.array([8, 2, 15, 6])
imax = np.argmax(numbers)
vmax = numbers[imax]

print(imax)
print(vmax)
```

```python
# What is the lowest value and its index?

imin = np.argmin(numbers)
vmin = numbers[imin]

print(imin)
print(vmin)
```

### argmax and argmin

| Code | Returns |
|---|---|
| `np.argmax(numbers)` | Index of the largest value |
| `np.argmin(numbers)` | Index of the smallest value |
| `numbers[i]` | Value at index `i` |

Richard’s lecture also explores how to find the maximum’s index using a loop, before comparing it with NumPy.

```python
# Loop to find the largest value

numbers = np.array([8, 2, 15, 6])

# This means: “Treat the first value (index [0]) as the largest found so far.” 
# We haven’t checked the remaining values yet.

imax = 0
vmax = numbers[imax]

# Produces indices 1, 2, 3, index 0 was already handled
for i in range(1, len(numbers)):
    # [i] is the index we're currently checking
    # Here we retrieve the value of that index
    v = numbers[i]
    # Checks whether this value beats the largest found so far
    if v > vmax:
        # Updates the index of the largest value
        imax = i
        # Updates the largest value
        vmax = v

print(imax)
print(vmax) 
print(np.argmax(numbers))
```

| Index checked (`i`) | Value (`v`) | `v > vmax`? | `imax` afterwards | `vmax` afterwards |
|---|---|---|---|---|
| 1 | 2 | False | 0 | 8 |
| 2 | 15 | True | 2 | 15 |
| 3 | 6 | False | 2 | 15 |

```python
# Try an example where two indexes hold the largest value

numbers = np.array([8, 15, 15, 6])

# Start with the first element as the largest found so far.
imax = 0
vmax = numbers[imax]

# Check each remaining element.
for i in range(1, len(numbers)):
    v = numbers[i]

    if v > vmax:
        imax = i
        vmax = v

print(imax)                # 1
print(vmax)                # 15
print(np.argmax(numbers))  # 1
```

Here, argmax keeps index 1, because 15 [2] > 15 [1] is False. 

np.argmax() also returns the first occurrence when the maximum appears more than once.

The loop tracks the largest value found so far and its index. It updates both whenever it finds a strictly larger value. NumPy’s argmax finds that index for us.

```python
# Let's look at an array with negative values

numbers = np.array([-8, -2, -5])

imax = 0
vmax = numbers[imax]  # Start with -8, an actual value in the array.

for i in range(1, len(numbers)):
    v = numbers[i]
    if v > vmax:
        imax = i
        vmax = v

print(imax)
print(vmax)
```

One clarification: starting with 0 doesn’t itself prevent checking the first element. Our loop skips index 0 because we already used that element to initialise vmax.

Start with the first element as the largest found so far. This works even when all values are negative. Then check the remaining elements.

Now let’s turn the loop into a reusable function, following Richard’s lecture:

```python
def argmax(values):
    N = len(values)

    # Check whether the input is empty
    if N == 0:
        # Stop loop with an explanatory error if there are no values in array
        print("Length zero sequences not support")
        return

    # Start with the first index as the largest found so far.
    imax = 0
    vmax = values[0]

    # Iterate through index 1 and the length of the array
    for i in range(1, N):
        # Check the value and its  index
        v = values[i]
        # If value is larger than previous value, set index as imax and value as vmax
        if v > vmax:
            imax = i
            vmax = v

    # The return keyword is to exit a function and return a value
    # Send the maximum’s index back to wherever the function was called
    return imax
    
```

```python
# Defining the function doesn’t run its body. Run that cell first, then call it:

numbers = np.array([8, 2, 15, 6])

position = argmax(numbers) # Calls on the function and receives the index: 2

print(position)
print(numbers[position]) # Retrieves the value: 15

# Directly use argmax, check against NumPy (same result)
print(np.argmax(numbers))
```

```python
# Check that our function doesn't support arrays with zero elements 

argmax([])
```
